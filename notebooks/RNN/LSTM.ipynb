```python
quote=df['quote']
quote.head()
```

```python
quote=quote.str.lower()
```

```python
import string
translator=str.maketrans('','',string.punctuation)
quote=quote.apply(lambda x: x.translate(translator))
```

```python
quote.head()
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocab_size = 10000

tokenizer = Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quote)
```

```python
word_index=tokenizer.word_index
print(len(word_index))
list(word_index.items())[:10]
```

```python
sequence = tokenizer.texts_to_sequences(quote)
```

```python
sequence[0]
```

```python
for i in range(3):
    print(sequence[i])
```

```python
x=[]
y=[]

for seq in sequence:
    for i in range(1,len(seq)):
        input_seq=seq[:i]
        output_seq=seq[i]
        x.append(input_seq)
        y.append(output_seq)

```

```python
x
```

```python
len(x)
```

```python
len(y)
```

```python
max_len=max(len(x) for x in x)
max_len
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
x_padded=pad_sequences(x,maxlen=max_len,padding='pre')
```

```python
x_padded
```

```python
y=np.array(y)
```

```python
x_padded.shape
```

```python
y.shape
```

```python
from tensorflow.keras.utils import to_categorical
y_one_hot=to_categorical(y,num_classes=vocab_size)
```

```python
y_one_hot.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,LSTM,Dense
```

```python
embedding_dim=50
rnn_units=128
```

```python
rnn_model = Sequential()

rnn_model.add(
    Embedding(
        input_dim=vocab_size,
        output_dim=embedding_dim,
        input_length=max_len
    )
)

rnn_model.add(SimpleRNN(units=rnn_units, return_sequences=True))
rnn_model.add(SimpleRNN(units=rnn_units))

rnn_model.add(Dense(units=vocab_size, activation="softmax"))
```

```python
rnn_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
rnn_model.summary()
```

# LSTM

```python
lstm_model = Sequential()

lstm_model.add(
    Embedding(
        input_dim=vocab_size,
        output_dim=embedding_dim,
        input_length=max_len
    )
)

lstm_model.add(SimpleRNN(units=rnn_units, return_sequences=True))
lstm_model.add(SimpleRNN(units=rnn_units))

lstm_model.add(Dense(units=vocab_size, activation="softmax"))
```

```python
lstm_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
lstm_model.summary()
```

```python
epochs=10
batch_size=128
```

```python
history_rnn=rnn_model.fit(
    x_padded,y_one_hot,
    epochs=epochs,
    batch_size=batch_size,
    validation_split=0.1
)
```
